```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("menu.csv")

print("First 5 Rows")
display(df.head())

print("Missing Values")
display(df.isnull().sum())

df.drop_duplicates(inplace=True)

print("Shape")
display(df.shape)
```

```python
print("Mean Calories:", df["Calories"].mean())
print("Median Calories:", df["Calories"].median())
print("Mode Calories:", df["Calories"].mode()[0])
print("Standard Deviation:", df["Calories"].std())
```

```python
category_calories = df.groupby("Category")["Calories"].mean()
print(category_calories)
```

```python
top_items = df.sort_values(by="Calories", ascending=False)

for i in range(10):
    print(top_items["Item"].iloc[i], "-", top_items["Calories"].iloc[i])
```

```python
category_count = df["Category"].value_counts()
print(category_count)
```

```python
category_count.plot(kind="bar")
plt.title("Number of Items in Each Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
category_calories.plot(kind="line")
plt.title("Average Calories by Category")
plt.xlabel("Category")
plt.ylabel("Calories")
plt.xticks(rotation=45)
plt.show()
```

```python
corr = df[["Calories", "Total Fat", "Carbohydrates", "Protein", "Sugars"]].corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
